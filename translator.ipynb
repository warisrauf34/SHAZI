## 2. Configure — API key, video URL, language

```python
import os

# @title Configuration { display-mode: 'form' }
GROQ_API_KEY = ''  # @param {type:'string'}
YOUTUBE_URL = 'https://www.youtube.com/watch?v=dQw4w9WgXcQ'  # @param {type:'string'}
TARGET_LANGUAGE = 'ur'  # @param ['ur', 'hi', 'en', 'es', 'ar', 'fr', 'de', 'pt', 'ru', 'tr', 'zh', 'ja']
VOICE = 'ur-PK-AsadNeural'  # @param ['ur-PK-AsadNeural', 'ur-PK-UzmaNeural', 'hi-IN-MadhurNeural', 'hi-IN-SwaraNeural', 'en-US-GuyNeural', 'en-US-JennyNeural', 'es-ES-AlvaroNeural', 'es-ES-ElviraNeural', 'ar-SA-HamedNeural', 'ar-SA-ZariyahNeural']

assert GROQ_API_KEY.strip(), 'Groq API key daalo! https://console.groq.com/keys se free milegi'
os.environ['GROQ_API_KEY'] = GROQ_API_KEY.strip()
print(f'✓ Configured for {TARGET_LANGUAGE} with voice {VOICE}')
```

## 3. Run the pipeline

5 steps: download → transcribe (Groq Whisper) → translate (Groq LLM) → TTS (Edge) → mux.

```python
from pathlib import Path
from translator.pipeline import run

output = run(
    url=YOUTUBE_URL,
    target_lang=TARGET_LANGUAGE,
    voice=VOICE,
    workdir=Path('/content/out'),
)
print(f'\n✓ Done: {output}')
```
